### MapPlot (Celestial Sphere)

https://starplotting.com/blog/celestial-sphere/

````python
''' Starplot
https://starplotting.com/blog/celestial-sphere/
'''

from starplot import (
    MapPlot,                      # ZenithPlot(), GalaxyPlot(),
    Mollweide,
    PlotStyle,
    style_extensions,             # extensions,
    _,
)

# --------------------------------

style = PlotStyle().extend(
    style_extensions.BLUE_NIGHT,  # .BLUE_MEDIUM, extensions.BLUE_NIGHT,
    style_extensions.MAP,
)

p = MapPlot(
    projection=Mollweide(),
    style=style,
    resolution=6000,
    autoscale=False,    # ?
    scale=0.8,
)

# --------------------------------

# decoration
p.gridlines(labels=False)
p.ecliptic(num_labels=3)
p.celestial_equator(num_labels=1)
p.milky_way()

# 88 Sternbilder
p.constellations()
#p.constellation_borders()
p.constellation_labels(style__font_size=20)

# definition for targets (others than Messier Objects)
mag = 10  # 12 for smart telescope, 6 for eye vision

# Sterne
p.stars(
    where=[_.magnitude < 4],        # mag < 6: visible by eye; mag < 4 shows only constellations
    where_labels=[_.magnitude < 3],
)

# Deep Sky Onjects (DSO) from catalogs
'''
p.messier(
    where_true_size=[False],
    label_fn=lambda d: f"M{d.m}",
)
'''
p.open_clusters(
    where=[_.magnitude < 6],
    where_labels=[True],
    where_true_size=[False],
)
p.globular_clusters(
    where=[_.magnitude < 6],
    where_labels=[True],
    where_true_size=[False],
)
p.galaxies(
    where=[_.magnitude < mag],
    where_labels=[True],
    where_true_size=[False],
)
p.nebula(
    where=[_.magnitude < mag],
    where_labels=[True],
    where_true_size=[False],
)

# --------------------------------

# Markers (POI)
style_marker={
    "marker": {
        "size": 60,
        "symbol": "circle",
        "fill": "none",
        "color": None,
        "edge_color": "hsl(44, 70%, 73%)",
        "edge_width": 2,
        "line_style": [1, [2, 3]],
        "alpha": 1,
        "zorder": 100,
    },
    "label": {
        "zorder": 200,
        "font_size": 22,
        "font_weight": 700,
        "font_color": "hsl(44, 70%, 64%)",
        "font_alpha": 1,
        "offset_x": "auto",
        "offset_y": "auto",
        "anchor_point": "top right",
    },
}

from astropy import units as u
from astropy.coordinates import SkyCoord, EarthLocation, AltAz

# targets for marker

targets = []    # list [] of dicts {} with keys 'catalog' and 'label'
targets.append({'catalog': 'M42',     'label': 'Orionnebel'})
targets.append({'catalog': 'M31',     'label': 'Andromeda Galaxie'})
targets.append({'catalog': 'M45',     'label': 'Plejaden'})
targets.append({'catalog': 'M44',     'label': 'Praesepe'})             # Praesepe | Bienenstockhaufen | Krippe'})# 
targets.append({'catalog': 'M13',     'label': 'Herkuleshaufen'})       # Herkuleshaufen | Great Hercules Cluster'})
targets.append({'catalog': 'M51',     'label': 'Whirlpool Galaxy'})
targets.append({'catalog': 'M81',     'label': 'Bodes Galaxie'})
targets.append({'catalog': 'M57',     'label': 'Ringnebel'})            # Ringnebel | NGC 6720'})
targets.append({'catalog': 'M101',    'label': 'Pinwheel Galaxy'})      # Pinwheel Galaxy | Feuerradgalaxie'})
targets.append({'catalog': 'M97',     'label': 'Owl Nebula'})
targets.append({'catalog': 'IC1805',  'label': 'Herznebel'})            # Herznebel | Heart Nebula'})
targets.append({'catalog': 'IC1848',  'label': 'Seelennebel'})          # Seelennebel | Soul Nebula'})
targets.append({'catalog': 'M17',     'label': 'Omega Nebel'})          # Omega Nebel | Schwanennebel | Swan Nebula'})
targets.append({'catalog': 'NGC7000', 'label': 'Nordamerika Nebel'})    # Nordamerika Nebel | North America Nebula'})
targets.append({'catalog': 'M65',     'label': 'Leo-Triplet'})          # Leo-Triplet (M65, M66, NGC3628)
targets.append({'catalog': 'NGC6992', 'label': 'Cirrusnebel'})          # Cirrusnebel | Schleiernebel | Veil Nebula'})
targets.append({'catalog': 'NGC281',  'label': 'Pacman-Nebel'})
#targets.append({'catalog': '',  'label': ''})

for target in targets:
    c = SkyCoord.from_name(target['catalog'])
    p.marker(
        ra=c.ra.degree,
        dec=c.dec.degree,
        style=style_marker,
        label=target['label']
    )

# --------------------------------

p.legend()
p.title("SkyMap with 88 constellations and a catalog of Deep-Sky-Objects (mag < 10, except clusters mag < 6, stars mag < 4)", style__font_size=36)  # 86

p.export("starplot.svg", format="svg", padding=0.25, transparent=True)

# --------------------------------

````
