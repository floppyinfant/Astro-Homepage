## Formeln

Der Zusammenhang zwischen Wellenlänge und Frequenz wird über die Ausbreitungsgeschwindigkeit der Welle beschrieben:

$c = \lambda \cdot \nu$

Einsteins Massen-Energie-Äquivalenz:

$E = mc^2$

Energie von Photonen:

$E = h \cdot \nu$

Impuls: 

$p = m \cdot v$

Die De Broglie Gleichung beschreibt den Welle-Teilchen-Dualismus:

$$\lambda = \frac{h}{p}$$

$$\lambda = \frac{h}{m \cdot v}$$


daraus folgt:

$$
\begin{aligned}
E &= m \cdot c^2 = h \cdot \nu && \quad \vert \div c \\
p &= m \cdot c = \frac{h \cdot \nu}{c} = \frac{h}{\lambda} \\
\\
&\text{mit:} \\
&\quad m: \text{Masse} \\
&\quad c: \text{Lichtgeschwindigkeit} \\
&\quad h: \text{Plancksches Wirkungsquantum} \\
&\quad \nu: \text{Frequenz} \\
&\quad \lambda: \text{Wellenlänge}
\end{aligned}
$$

## Objektive

### Brennweite (f)

Barlow-Linse: verdoppelt die Brennweite, aber auch die f-Zahl (das Bild wird 4 mal dunkler)

### Blende, aperture

#### Öffnungsverhältnis

$$f/ = \frac{Brennweite}{Öffnung (aperture)}$$

f/2 bis f/5: schnelles System, sammelt viel Licht in kurzer Zeit

f/10 bis f/15: langsames System (z.B. Schmidt-Casegrains) werden für helle Planeten genutzt

### Vergrößerung

$$Vergrößerung = \frac{Teleskop-Brennweite}{Okular-Brennweite}$$

In der Praxis ist die Vergrößerung durch den Objektivdurchmesser (Trennschärfe, Auflösung) und die Luftunruhe (Seeing) begrenzt.

Fernglas Angaben: Vergrößerung X Durchmesser der Linse
- 8 x 40
- 8 x 50
- 10 x 50

#### Auflösung, Trennschärfe

Faustformel

$$max. Vergrößerung = 2 \cdot Objektivdurchmesser$$

$$min. Vergrößerung = \frac{Teleskop-Objektivdurchmesser}{Pupillendurchmesser}$$

Beispiel 1: Objektiv-Durchmesser = 60 mm
- kann Sterne mit einem Abstand von 3" trennen
- Vergrößerung min. 12-fach, max. 120-fach

Beispiel 2: Objektiv-Durchmesser 120 mm
- Trennschärfe 1,5"
- min. Vergrößerung 24-fach
- max. Vergrößerung 240-fach

Die Auflösung der Luftunruhe (Seeing) beträgt meist 2".

### Field of View (FOV)

Syn: Bildwinkel, Blickfeld, Sichtfeld, Gesichtsfeld

$$
\begin{aligned}
\alpha &= 2 \cdot \arctan\left(\frac{d}{2 \cdot f}\right) \\[0.8em]
&\text{mit:} \\
&\quad \alpha: \text{Bildwinkel (in Grad)} \\
&\quad d: \text{Sensorgröße (z. B. Breite des Sensors in mm)} \\
&\quad f: \text{Brennweite des Objektivs (in mm)} \\
&\quad \arctan: \text{Arkustangens (die Umkehrfunktion des Tangens)}
\end{aligned}
$$

Ergebnis in Radiant. Umrechnung in Grad:

$$\alpha_{\text{Grad}} = \alpha_{\text{Rad}} \cdot \frac{360^\circ}{2 \pi} \approx \alpha \cdot 57,3°$$



### Auflösung

### Pixelgröße

pixel-pitch

$$Pixelgröße\ [\mu m] = \frac{Sensorbreite\ [mm]}{Anzahl der horizontalen Pixel} \cdot 1000$$

### Abbildungsmaßstab

image scale, pixel scale

Der Abbildungsmaßstab (pixel scale) gibt an, wieviele Bogensekunden (arcseconds) des Himmels ein einelnes Pixel abdeckt.

$$pixel scale\ ["/px] = \frac{Pixelgröße\ [\mu m] \cdot 206.265}{Brennweite\ [mm]}$$

Oversampling: < 1"/px

Die Pixel sind sehr klein im Vergleich zur Luftunruhe (Seeing).

Undersampling: > 2"/px

Ein pixel deckt einen großen Bereich ab. Sterne können blockig wirken.
